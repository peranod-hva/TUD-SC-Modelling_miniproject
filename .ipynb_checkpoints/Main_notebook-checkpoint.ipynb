# 1.4: Drawing shapes

We start by importing our prior solutions into Python:

```python
import numpy as np
import matplotlib.pyplot as plt

# Dummy constants, to be replaced with reasonable order of magnitude estimates:
r_m = 1
L_m = 1
R_m = 1
J_m = 1
K_m = 1
m = 1

A = np.array([[0,             0, r_m*K_m / (2*J_m + m*r_m**2))],
              [1,             0,              0],
              [0,             0,         -R_m / L_m]])

B = np.array([[0], [0], [1/L_m]])
```

To draw an ellipsoid, we define a trajectory with a semi-x-axis and a semi-y-axis:

```python
semi_xaxis = 2
semi_yaxis = 1
```

It then follows that
x = semi_xaxis cos(t)
y = semi_yaxis sin (t)
0 <= t < 2pi

x' = -semi_xaxis sin(t)
x'' = -semi_xaxis cos(t)
x''' = semi_xaxis sin(t)

y' = semi_yaxis cos (t)
y'' = -semi_yaxis sin (t)
y''' = -semi_yaxis cos (t)
